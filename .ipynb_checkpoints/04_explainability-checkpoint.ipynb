```python
# Phase 5: SHAP Explainability

Accuracy tells us HOW WELL the model predicts.
SHAP tells us WHY it makes each prediction.

This is what turns our model from a black box into
an actionable business tool.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pickle
import shap

print("SHAP version:", shap.__version__)
```

```python
# Load saved model
with open('churn_model.pkl', 'rb') as f:
    model = pickle.load(f)

# Load test data
X_test  = pd.read_csv('X_test.csv')
y_test  = pd.read_csv('y_test.csv').squeeze()
X_train = pd.read_csv('X_train.csv')

print(f"Model loaded ✓")
print(f"X_test shape : {X_test.shape}")
print(f"X_train shape: {X_train.shape}")
```

```python
# TreeExplainer is optimized for XGBoost — fast and accurate
explainer = shap.TreeExplainer(model)

# Calculate SHAP values for test set
# This is the core step — may take 10-20 seconds
shap_values = explainer(X_test)

print(f"SHAP values shape: {shap_values.shape}")
print(f"\nFor each of the {X_test.shape[0]} customers,")
print(f"we now have a SHAP value for each of the {X_test.shape[1]} features.")
print("\nSHAP values calculated ✓")
```

```python
# This is the most important plot in your entire project
# Shows which features drive churn across ALL customers

plt.figure(figsize=(10, 8))
shap.plots.beeswarm(shap_values, max_display=15, show=False)
plt.title('SHAP Beeswarm — Top 15 Features Driving Churn',
          fontsize=13, pad=12)
plt.tight_layout()
plt.savefig('shap_beeswarm.png', dpi=150, bbox_inches='tight')
plt.show()
print("\nHow to read this plot:")
print("- Features ranked by importance (top = most important)")
print("- RED dots = high feature value, BLUE = low feature value")
print("- Dots on RIGHT = pushes toward churn")
print("- Dots on LEFT  = pushes away from churn")
```

```python
# Simpler version — just shows average impact of each feature
plt.figure(figsize=(10, 7))
shap.plots.bar(shap_values, max_display=15, show=False)
plt.title('SHAP Feature Importance — Average Impact on Churn',
          fontsize=13, pad=12)
plt.tight_layout()
plt.savefig('shap_bar.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Find a customer with HIGH churn probability
probs = model.predict_proba(X_test)[:, 1]
high_risk_idx = probs.argmax()  # highest churn probability

print(f"Highest risk customer — index: {high_risk_idx}")
print(f"Churn probability: {probs[high_risk_idx]*100:.1f}%")
print(f"Actual outcome: {'Churned' if y_test.iloc[high_risk_idx]==1 else 'Did not churn'}")
print(f"\nTop features for this customer:")
print(X_test.iloc[high_risk_idx][
    abs(shap_values[high_risk_idx].values).argsort()[::-1][:5]
])
```

```python
# This chart explains WHY this specific customer is at high risk
# This is what you'll show in your dashboard and LinkedIn post

plt.figure(figsize=(10, 6))
shap.plots.waterfall(shap_values[high_risk_idx], 
                     max_display=12, show=False)
plt.title(f'Why is this customer at risk? (Churn prob: {probs[high_risk_idx]*100:.1f}%)',
          fontsize=12, pad=12)
plt.tight_layout()
plt.savefig('shap_waterfall_highrisk.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nHow to read this plot:")
print("- Bars going RIGHT (red)  = pushing toward churn")
print("- Bars going LEFT (blue)  = pushing away from churn")
print("- E[f(x)] = average prediction across all customers")
print("- f(x)    = final prediction for THIS customer")
```

```python
# Compare with a LOW risk customer — great for storytelling
low_risk_idx = probs.argmin()

print(f"Lowest risk customer — index: {low_risk_idx}")
print(f"Churn probability: {probs[low_risk_idx]*100:.1f}%")
print(f"Actual outcome: {'Churned' if y_test.iloc[low_risk_idx]==1 else 'Did not churn'}")

plt.figure(figsize=(10, 6))
shap.plots.waterfall(shap_values[low_risk_idx],
                     max_display=12, show=False)
plt.title(f'Why is this customer safe? (Churn prob: {probs[low_risk_idx]*100:.1f}%)',
          fontsize=12, pad=12)
plt.tight_layout()
plt.savefig('shap_waterfall_lowrisk.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Convert SHAP values into a readable dataframe
# This gets used in the business metrics and dashboard later

shap_df = pd.DataFrame(
    shap_values.values,
    columns=X_test.columns
)

# Get top 5 risk factors per customer
def get_top_risk_factors(row, n=3):
    top = row.abs().nlargest(n).index.tolist()
    return ', '.join(top)

shap_df['top_risk_factors'] = shap_df.apply(get_top_risk_factors, axis=1)

# Add churn probability
shap_df['churn_probability'] = probs

# Save for dashboard use later
shap_df.to_csv('shap_values.csv', index=False)
X_test['churn_probability'] = probs
X_test['top_risk_factors']  = shap_df['top_risk_factors'].values
X_test['actual_churn']      = y_test.values
X_test.to_csv('X_test_with_shap.csv', index=False)

print("SHAP dataframe saved ✓")
print(f"\nSample — top risk factors per customer:")
print(shap_df[['churn_probability', 'top_risk_factors']].head(10))
```

```python
## SHAP Analysis Findings

### Global Insights (What drives churn overall?)

Based on the beeswarm plot, the top churn drivers are:

1. **Contract type (month-to-month)** — single biggest driver.
   Customers on monthly contracts have dramatically higher churn risk.

2. **Tenure** — low tenure strongly pushes toward churn.
   New customers need immediate attention and onboarding.

3. **Monthly charges** — higher charges increase churn risk.
   Customers who pay more but use fewer services feel poor value.

4. **Internet service (Fiber optic)** — surprisingly high churn signal.
   Fiber optic customers churn more — possible service quality issue.

5. **No online security / tech support** — customers without
   support services have less reason to stay.

### Local Insights (Why does THIS customer churn?)

The waterfall plots show that for individual customers, churn risk
is driven by a unique combination of factors — not just one feature.
This is why per-customer SHAP explanations are more actionable than
global feature importance alone.

### Business Recommendation

The highest risk segment is:
- Month-to-month contract
- Tenure < 12 months  
- Monthly charges > $65
- No online security or tech support

These customers should be prioritized for retention offers immediately.
```
